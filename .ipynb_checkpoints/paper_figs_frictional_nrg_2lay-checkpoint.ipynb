```julia
# loading modules
using PyPlot, Printf, LinearAlgebra, FFTW, FileIO, JLD2, Statistics

include("../../LinStab/mjcl_stab.jl")
using .LinStab

using FourierFlows: CPU, TwoDGrid

using GeophysicalFlows

using Glob

using FileIO, JLD2
```

```julia
# numerical params
Ny = Nx = n = 1024              # 2D resolution = n²
stepper = "FilteredRK4"         # time stepping scheme

dev = CPU()

# physical params

nlayers = 2                 # number of layers
f0 = f₀= 8.3e-5            # Coriolis param [s^-1]
g = 9.81                    # gravity
H = [2000., 2000.]     # the rest depths of each layer;

rho0 = 1025.        # Boussinesq reference density

# always zero for now, model can't even take this in as a parameter...
V = zeros(nlayers)

# Linear bottom drag
μ = (86400*45)^-1

# running model
Us = [0.01, 0.017634058790440166, 0.024]

rhos = 1025.75 - 0.5775


rho = ρ = [0., 1025.75]
rho[1] = ρ[1] = rhos[1]
U = [0.,0.]

U[1] = Us[1]

gp = (g/rho0)*(ρ[2] - ρ[1])

dt=3600;

Ld = sqrt(gp*sum(H))/f0/2

L = Lx = Ly = 25*2*pi*Ld   # domain size [m]
beta = β = 0. * (U[1]/2) * Ld^-2         # 1e-11              # the y-gradient of planetary PV


# TY10: kappa^* = 0.1 (mu=328^-1), beta^* = 0.75, L=64*L_d
# TY07: kappa^* = 0.08 (mu=410^-1), beta^* = 0.75, L=25*2*pi*L_d


U[1] = 0.01 # Us[1]


# domain size [m]

global ss_yr_max = 1000 * (Ld / U[1]) / 3600 / 24 / 365.25

global dt = 0.
global nsubs = 0.
global cfl_glob = 0.5   # 0.01 is standard so far; 0.005 for h[1] rho[1] only
global dx = L/Nx
# setting cfl and dt
global dt = (Ld/Nx)*cfl_glob/U[1]    # divisor of 5 bc 5x increase in L_x
# global dt = dt/2      # time step change for mu 500 and 1000
# global dt = dt/2      # time step change for mu1000, h0s 6 and 7

# sampling period in time steps (set above)

S32 = f0 * U[1] / gp
dyetab_over_S32 = collect(range(-1.5,1.5,13))
h0s = dyetab_over_S32 .* S32

# biharmonic viscosity
global nν = 4
global ν  = 10^-14 * (Us[1]/2) * (Lx/2/pi)^7

topographic_pv_gradient = (0., h0s[1]*(f0/H[end]))



prob = MultiLayerQG.Problem(nlayers, dev; nx=n, Lx=L, f₀, H, g, ρ, U, nν, ν, topographic_pv_gradient,
μ, β, dt, stepper, aliased_fraction=1/3);

sol, clock, params, vars, grid_jl = prob.sol, prob.clock, prob.params, prob.vars, prob.grid;

```

```julia
xh = sqrt.( 2 * kr.^2) *Ld

function argmax_TT(TT)
    return [xh[argmax(((x-7)/abs(x-7)) * TT[:, x, 2], dims=1)[1]] for x in collect(range(1,size(TT)[2]))]
end

function argmax_DBT(DBT)
    return [xh[argmin(DBT[:, x, 1], dims=1)[1]] for x in collect(range(1,size(DBT)[2]))]
end

```

```julia
function iso_spec(u, v, grid)
    
    uk = abs.(rfft(u .- mean(u))) .* (2/grid.nx)
    vk = abs.(rfft(v .- mean(v))) .* (2/grid.nx)
    
    dk = 2*pi/grid.Lx; dl = 2*pi/grid.Ly;
    
    dkr = sqrt(dk^2 + dl^2)
    
    wv = @. sqrt(grid.kr^2 + grid.l^2)
    
    ispec = zeros(length(grid.kr))
    
    for i in range(1,length(grid.kr))
        # find 2D index values for a wavenumber magnitude
        if i==length(grid.kr)
            fkr = @. (wv>=grid.kr[i]) & (wv<=grid.kr[i]+dkr)
        else
            fkr = @. (wv>=grid.kr[i]) & (wv<grid.kr[i+1])
        end
        
        if sum(fkr) > 0

            ispec[i] = wv[i]^2 .* sum(uk[fkr].^2 .+ vk[fkr].^2) # this is energy spectrum

        end
        
    end

    
    return ispec
    
end
```

```julia
using FileIO, JLD2, FFTW, Statistics

function zonal_hov(U,ρ,H,μ,h0s,Lx,k)
    nx=512
    
    topographic_pv_gradient = (0, (f₀/H[end]) * h0s[k])
        
    prob = MultiLayerQG.Problem(nlayers, dev; nx=Nx, Lx=Lx, f₀, H, g, ρ, U, topographic_pv_gradient,
    μ, β, dt, stepper, aliased_fraction=1/3);

    sol, clock, params, vars, grid = prob.sol, prob.clock, prob.params, prob.vars, prob.grid;

    GC.gc()

    file1 = jld_name(data_dir,h0s[k],Lx,ρ,U,Nx,μ,H,1)
    files = glob(file1[52:end-5]*"*", data_dir)

    N_yr = length(files)
    yrs = collect(range(1,N_yr))
        
    ## pre-allocate
    window = N_yr-1 # analyze last 20 years
    nmax = window*13*60  # assume 13 outputs per year, 60 outputs per domain sweep (see step in st) i.e., more than possible with monthly output
    
    hov0 = zeros(nmax,nx)

    n=1
    ##

    for i in range(N_yr-window+1,N_yr)

        a = load(jld_name(data_dir,h0s[k],Lx,ρ,U,Nx,μ,H,i))
        # a = load(jld_name(data_dir,h0s[h_ind],Lx,rho,U,Nx,μ,H,N_yr))

        s = size(a["jld_data"]["psi_ot"])[4]

        for j in range(1,s)
            
            psi = a["jld_data"]["psi_ot"][:,:,:,j]

            ψ = 0.5 * (psi[:,:,1] .+ psi[:,:,2])
            
            ψh = rfft(ψ)
            
            ∂yψh = im * grid.l .* ψh
            
            ∂yψ = irfft(∂yψh, grid.ny)
                                
            hov0[n,:] = mean(-∂yψ, dims=1)

            # # new code for Reynolds stresses
            # u1 = irfft(-im * grid_jl.l  .* rfft(psi[:,:,1]), grid_jl.ny)                
            # v1 = irfft( im * grid_jl.kr .* rfft(psi[:,:,1]), grid_jl.ny)

            # u1v1_bar = mean(u1 .* v1, dims=1)
            
            # d_uBTbar_dy = irfft(im * grid.l .* rfft(mean(-∂yψ, dims=1)),1)

            # rs0[n,:] = d_uBTbar_dy .* u1v1_bar

            n+=1
        end

        GC.gc()


    end
    gp = (g/rho0)*(ρ[2] - ρ[1])
    
    Ld = sqrt(gp*sum(H))/f0/2
    
    t = collect(range(0,(n-1)*61*24*3600,n-1)) .* U[1] ./ Ld 
    return t, hov0[1:n-1,:]
end
```

# Wavenumber spectrum as f(t) sandbox

```julia
using Statistics

function isotropic_mean(arr_in, grid)
    # arr_in: an nkr X nl array that is output of rfft
    # note that we only want real part of this
    
    dk = 2*pi/grid.Lx; dl = 2*pi/grid.Ly;
    
    dkr = sqrt(dk^2 + dl^2)
    
    wv = @. sqrt(grid.kr^2 + grid.l^2)
    
    iso = zeros(length(grid.kr))
    
    for i in range(1,length(grid.kr))
        # find 2D index values for a wavenumber magnitude
        if i==length(grid.kr)
            fkr = @. (wv>=grid.kr[i]) & (wv<=grid.kr[i]+dkr)
        else
            fkr = @. (wv>=grid.kr[i]) & (wv<grid.kr[i+1])
        end
        
        if sum(fkr) > 0
            iso[i] = mean(real(arr_in[fkr])) # this is average over all combinations of k_x and k_y that are the same, isotropic k
        end
        
    end

    return iso
end
```

```julia
function modal_spec_budget(vars, params, grid_jl, sol, ψ,μstar, νstar, drag_bool)

    # parameters
    nlayers = 2
    δ = [params.H[1]/sum(params.H), params.H[2]/sum(params.H)]
    g′ = params.g * (params.ρ[2] -   params.ρ[1]) / 1025. # reduced gravity at each interface
    Ld1 = (params.f₀^2 / (g′ * params.H[1]))^-0.5
    ∂yηb = params.topographic_pv_gradient[2] / (params.f₀ / params.H[end])
    
    # assigning basic variables
    vars.ψ .= ψ
    # vars.q .= q
    
    vars.ψh[:,:,1] .= rfft(vars.ψ[:,:,1])
    vars.ψh[:,:,2] .= rfft(vars.ψ[:,:,2])

    ψBC = 0.5 * (vars.ψ[:,:,1] .- vars.ψ[:,:,2])
    ψBT = 0.5 * (vars.ψ[:,:,1] .+ vars.ψ[:,:,2])
    
    ψBCh = rfft(ψBC)
    ψBTh = rfft(ψBT)
    
    U₁, U₂, = view(params.U, :, :, 1), view(params.U, :, :, 2)
    
    S32 = params.f₀ * U₁[1,1] / g′

    # streamfunction stuff
    ∂xψBTh = im * grid_jl.kr .* ψBTh
    ∂yψBTh = im * grid_jl.l .* ψBTh
    
    ∂xψBT = irfft(∂xψBTh, grid_jl.ny)
    ∂yψBT = irfft(∂yψBTh, grid_jl.ny)

    ∂xψBCh = im * grid_jl.kr .* ψBCh
    ∂yψBCh = im * grid_jl.l .* ψBCh
    
    ∂xψBC = irfft(∂xψBCh, grid_jl.ny)
    ∂yψBC = irfft(∂yψBCh, grid_jl.ny)
    
    # nonlinear terms
    ζ = deepcopy(vars.v)
    ζ[:,:,1] .= irfft(-grid_jl.Krsq .* vars.ψh[:,:,1], grid_jl.ny)
    ζ[:,:,2] .= irfft(-grid_jl.Krsq .* vars.ψh[:,:,2], grid_jl.ny)
    ζ₁, ζ₂ = view(ζ, :, :, 1), view(ζ, :, :, 2)

    ζBCh = - grid_jl.Krsq .* ψBCh
    ζBTh = - grid_jl.Krsq .* ψBTh

    ζBC = irfft(ζBCh,grid_jl.ny)
    ζBT = irfft(ζBTh,grid_jl.ny)

    ∂xζBCh = im * grid_jl.kr .* ζBCh
    ∂xζBTh = im * grid_jl.kr .* ζBTh

    ∂yζBCh = im * grid_jl.l .* ζBCh
    ∂yζBTh = im * grid_jl.l .* ζBTh

    ∂xζBC = irfft(∂xζBCh, grid_jl.ny)
    ∂xζBT = irfft(∂xζBTh, grid_jl.ny)

    ∂yζBC = irfft(∂yζBCh, grid_jl.ny)
    ∂yζBT = irfft(∂yζBTh, grid_jl.ny)
    
    ζBT∂xψBTh = rfft(ζBT .* ∂xψBT)
    ζBT∂yψBTh = rfft(ζBT .* ∂yψBT)

    ζBC∂xψBCh = rfft(ζBC .* ∂xψBC)
    ζBC∂yψBCh = rfft(ζBC .* ∂yψBC)

    ζBC∂xψBTh = rfft(ζBC .* ∂xψBT)
    ζBC∂yψBTh = rfft(ζBC .* ∂yψBT)

    ζBT∂xψBCh = rfft(ζBT .* ∂xψBC)
    ζBT∂yψBCh = rfft(ζBT .* ∂yψBC)

    ψBC∂xψBT = rfft(ψBC .* ∂xψBT)
    ψBC∂yψBT = rfft(ψBC .* ∂yψBT)

    J_ψBT_ζBT = ∂xψBT .* ∂yζBT .- ∂yψBT .* ∂xζBT

    J_ψBT_ζBTh = rfft(J_ψBT_ζBT)

    ζ₁h = rfft(ζ₁)
    
    ζ₂h = rfft(ζ₂)
    
    ∂xζ1h = im * grid_jl.kr .* ζ₁h 
    ∂xζ1 = irfft(∂xζ1h, grid_jl.ny)

    ############################################################################################
    BTEKE = @. 0.5 * (conj(ψBTh) * (grid_jl.kr^2 * ψBTh) + ψBTh * conj(grid_jl.kr^2 * ψBTh))

    BCEKE = @. 0.5 * grid_jl.kr^2 * (conj(ψBCh) * ψBCh + ψBCh * conj(ψBCh))
    BCEAPE = @. 0.5 * 2 * Ld1^-2 * (conj(ψBCh) * ψBCh + ψBCh * conj(ψBCh))

    ############################################################################################
    # Baroclinic conversion term
    CBC = @. S32 * (params.f₀ / params.H[1]) * ( conj(ψBCh) * ∂xψBTh + ψBCh * conj(∂xψBTh)) # these are equivalent!

    # CBC = @. U₁ * Ld1^-2 * (conj(ψBCh) * ∂xψBTh)
    # CBC .+= conj(CBC)
    
    ############################################################################################
    # Linear BC-BT flux (flat-bottom) term
    BC2BT = @. 0.5 * U₁ * (conj(ψBTh) * ∂xζ1h + ψBTh * conj(∂xζ1h))

    ############################################################################################
    # Nonlinear terms in BT budget
    NLBT = zeros(grid_jl.nkr,grid_jl.ny,2) .+ 0im
    NLBT[:,:,1] = @. conj(ψBTh) * J_ψBT_ζBTh + ψBTh * conj(J_ψBT_ζBTh)  # im * (grid_jl.l * ζBT∂xψBTh - grid_jl.kr * ζBT∂yψBTh)
    # NLBT[:,:,1] .+= conj.(NLBT[:,:,1])

    NLBT[:,:,2] = @. conj(ψBTh) * im * (grid_jl.l * ζBC∂xψBCh - grid_jl.kr * ζBC∂yψBCh)
    NLBT[:,:,2] .+= conj.(NLBT[:,:,2])

    ############################################################################################
    # Nonlinear terms in BC budget
    NLBC = zeros(grid_jl.nkr,grid_jl.ny,3) .+ 0im
    NLBC[:,:,1] = @. conj(ψBCh) * im * (grid_jl.l * ζBC∂xψBTh - grid_jl.kr * ζBC∂yψBTh)
    NLBC[:,:,1] .+= conj.(NLBC[:,:,1])

    NLBC[:,:,2] = @. conj(ψBCh) * im * (grid_jl.l * ζBT∂xψBCh - grid_jl.kr * ζBT∂yψBCh)
    NLBC[:,:,2] .+= conj.(NLBC[:,:,2])

    NLBC[:,:,3] = @. - 2 * Ld1^-2 * conj(ψBCh) * im * (grid_jl.l * ψBC∂xψBT - grid_jl.kr * ψBC∂yψBT)
    NLBC[:,:,3] .+= conj.(NLBC[:,:,3])

    ############################################################################################
    TopoT = @. -0.5 * (params.f₀ / params.H[2]) * ∂yηb * (conj(ψBTh) * ∂xψBCh + ψBTh * conj(∂xψBCh))

    ############################################################################################
    # BT drag
    DBT = @. - 0.5 * params.μ * (conj(ψBTh) * (ζBCh - ζBTh) + ψBTh * conj(ζBCh - ζBTh))

    ############################################################################################
    # BC drag
    DBC = @. 0.5 * params.μ * (conj(ψBCh) * (ζBCh - ζBTh) + ψBCh * conj(ζBCh - ζBTh) )

    ############################################################################################
    # taking mean at each wavenumber magnitude, i.e., assuming isotropy
    NRGs = hcat(isotropic_mean(BTEKE,grid_jl), isotropic_mean(BCEKE,grid_jl), isotropic_mean(BCEAPE,grid_jl))
    CBCh = isotropic_mean(CBC, grid_jl)
    LF = hcat(isotropic_mean(BC2BT, grid_jl), isotropic_mean(TopoT, grid_jl))
    Drag = hcat(isotropic_mean(DBT, grid_jl), isotropic_mean(DBC, grid_jl))
    NLBTh = hcat(isotropic_mean(NLBT[:,:,1], grid_jl), isotropic_mean(NLBT[:,:,2], grid_jl))
    NLBCh = hcat(isotropic_mean(NLBC[:,:,1], grid_jl), isotropic_mean(NLBC[:,:,2], grid_jl), isotropic_mean(NLBC[:,:,3], grid_jl))

    NRGs = dropdims(NRGs,dims=tuple(findall(size(NRGs).==1)...))
    CBCh = dropdims(CBCh,dims=tuple(findall(size(CBCh).==1)...))
    LF = dropdims(LF,dims=tuple(findall(size(LF).==1)...))
    Drag = dropdims(Drag,dims=tuple(findall(size(Drag).==1)...))
    NLBTh = dropdims(NLBTh,dims=tuple(findall(size(NLBTh).==1)...))
    NLBCh = dropdims(NLBCh,dims=tuple(findall(size(NLBCh).==1)...))

    resid = CBCh .+ sum(Drag, dims=2) .+ sum(NLBTh, dims=2) .+ sum(NLBCh, dims=2)
    
    GC.gc()
   
  return NRGs, CBCh, LF, Drag, NLBTh, NLBCh, resid
end

modal_spec_budget(prob, ψ, μstar, νstar, drag_bool) = modal_spec_budget(prob.vars, prob.params, prob.grid, prob.sol, ψ,μstar, νstar, drag_bool)
```

```julia

function spec_budget(U,ρ,H,μ,h0,μstar, νstar, drag_bool)

    nlayers=2
    topographic_pv_gradient = (0, (f₀/H[end]) * h0s[1])

    μ = μstar * (U[1]/2) / Ld
        
    prob = MultiLayerQG.Problem(nlayers, dev; nx=Nx, Lx=L, f₀, H, g, ρ, U, topographic_pv_gradient,
    μ, β, dt, stepper, aliased_fraction=1/3);

    sol, clock, params, vars, grid_jl = prob.sol, prob.clock, prob.params, prob.vars, prob.grid;
        
    topographic_pv_gradient = (0, (f₀/H[end]) * h0)
        
    prob = MultiLayerQG.Problem(nlayers, dev; nx=Nx, Lx=L, f₀, H, g, ρ, U, topographic_pv_gradient,
    μ, β, dt, stepper, aliased_fraction=1/3);

    sol, clock, params, vars, grid_jl = prob.sol, prob.clock, prob.params, prob.vars, prob.grid;

    GC.gc()

    file1 = jld_name2(data_dir,μstar, νstar, drag_bool, L, h0,ρ,U,Nx,H,1)
    files = glob(file1[34:end-5]*"*", data_dir)

    if μstar<1.55
        N_yr=90
    else
        N_yr = 90 # length(files)
    end
        
    ## pre-allocate
    window = Int(round(250 * (Ld / U[1]) / 3600 / 24 / 365.25))   # analyze last 500 eddy-turnover times
    nmax = window*13  # assume 13 outputs per year, i.e., more than possible with monthly output
    
    NRGs_h0 = zeros(grid_jl.nkr, nmax, 3)
    CBC_h0 = zeros(grid_jl.nkr, nmax, nlayers-1)
    LF_h0 = zeros(grid_jl.nkr, nmax,2)
    Drag_h0 = zeros(grid_jl.nkr, nmax,2)
    NLBT_h0 = zeros(grid_jl.nkr, nmax, 2)
    NLBC_h0 = zeros(grid_jl.nkr, nmax, 3)
    resid_h0 = zeros(grid_jl.nkr, nmax)
    te = zeros(nmax,)

    n=1
    ##

    for i in range(2,50)

        a = load(jld_name2(data_dir,μstar, νstar, drag_bool, L, h0,ρ,U,Nx,H,i))

        s = size(a["jld_data"]["psi_ot"])

        if length(s)>3
            s=s[4]
        else
            s=1
        end

        for j in range(1,s)
            ψ = a["jld_data"]["psi_ot"][:,:,:,j]

            NRGs_h0[:,n,:], CBC_h0[:,n], LF_h0[:,n,:], Drag_h0[:,n,:], NLBT_h0[:,n,:], NLBC_h0[:,n,:], resid_h0[:,n] = modal_spec_budget(prob, ψ,μstar, νstar, drag_bool);

            te[n] = a["jld_data"]["t"][j]* (U[1]/2/Ld)
            
            n+=1
        end
        GC.gc()
    end

    NRGs_out =NRGs_h0[:,1:n-1,:]
    CBC_out = CBC_h0[:,1:n-1]
    LF_out = LF_h0[:,1:n-1,:]
    Drag_out = Drag_h0[:,1:n-1,:]
    NLBT_out = NLBT_h0[:,1:n-1,:]
    NLBC_out = NLBC_h0[:,1:n-1,:]
    resid_out = resid_h0[:,1:n-1]
    te_out = te[1:n-1]
    
    return NRGs_out, CBC_out, LF_out, Drag_out, NLBT_out, NLBC_out, resid_out, te_out
end
```

```julia
using Statistics

rho[1] = rhos[1]

rho0=1025.
U[1] = Us[1]

# U[1] = 0.015
# rho[1] = 1025.75-1.3

gp = (g/rho0)*(ρ[2] - ρ[1])



S32 = f0 * U[1] / gp
dyetab_over_S32 = collect(range(-1.5,1.5,16))
h0s = dyetab_over_S32 .* S32
h0s = [h0s[1:8]; 0.; h0s[9:end]]

h_ind = 9


# h0s = 3*S32

H=[2000., 2000.]

# Ld = sqrt(gp * sum(H))/f0 

# for i in range(2,N_yr)
#     for j in range(1,12)


topographic_pv_gradient = (0., h0s[9]*(f0/H[end]))

nν=10
ν= 2.5e46

prob = MultiLayerQG.Problem(nlayers, dev; nx=n, Lx=L, f₀, H, g, ρ, U, nν, ν, topographic_pv_gradient,
μ, β, dt, stepper, aliased_fraction=1/3);

sol, clock, params, vars, grid_jl = prob.sol, prob.clock, prob.params, prob.vars, prob.grid;


function plot_nrg_terms(mu, h_ind, N_yr, ax1, ax2, xl, yl, levs, cont_bool)
    
    a = load(jld_name(data_dir,h0s[h_ind],Lx,rho,U,Nx,mu,H,N_yr))


    ψtot = psi = a["jld_data"]["psi_ot"]
    
    t_ind = 1
    
    τ = 0.5 * (ψtot[:,:,1,t_ind] .- ψtot[:,:,2,t_ind])
    ψ = 0.5 * (ψtot[:,:,1,t_ind] .+ ψtot[:,:,2,t_ind])
    
    τh = rfft(τ)
    ψh = rfft(ψ)
    
    ∂xτh = im * grid_jl.kr .* τh
    ∂xψh = im * grid_jl.kr .* ψh
    
    ∂yτh = im * grid_jl.l .* τh
    ∂yψh = im * grid_jl.l .* ψh
    
    ∂xτ = irfft(∂xτh, grid_jl.ny)
    ∂xψ = irfft(∂xψh, grid_jl.ny)
    
    ∂yτ = irfft(∂yτh, grid_jl.ny)
    ∂yψ = irfft(∂yψh, grid_jl.ny)

    ∂xψ2 = ∂xψ .- ∂xτ

    u1 = ∂yψ .+ ∂yτ
    u2 = ∂yψ .- ∂yτ

    v1 = ∂xψ .+ ∂xτ
    v2 = ∂xψ .- ∂xτ

    BCprod = 0.01 * (28325)^-2 * τ .* ∂xτ
    
    BCtrans = (f0/H[1]) * S32 * τ .* ∂xψ2
    
    T_topo = (f0/ 2 / H[2]) * h0s[h_ind] * τ .* ∂xψ2

    T_topo_domain_avg = (f0/ 2 / H[2]) * h0s[h_ind] * τ .* ∂xψ

    # ∇ψdot∇τ = ∂xτ .* ∂xψ .+ ∂yτ .* ∂yψ
    
    # ∇ψ2 = ∂xψ.^2 .+ ∂yψ.^2
    
    ∇τ2 = ∂xτ.^2 .+ ∂yτ.^2
    
    ∇τdot∇ψ = ∂xτ .* ∂xψ .+ ∂yτ .* ∂yψ
    
    DEAPE = -(μ/2) * (∇τ2 - ∇τdot∇ψ)

    tot_trans = (f0/H[1] * S32 + f0 / 2 / H[2] * h0s[h_ind]) * τ .* ∂xψ2

    ##############################################################################3
    # plot_var1 = DEAPE    # 
    plot_var1 = BCprod
    # plot_var1 = BCtrans
    # plot_var1 = T_topo_domain_avg
    # plot_var1 = tot_trans
    # plot_var1 = u1
    
    norm3 = matplotlib.colors.TwoSlopeNorm(vcenter=0.,vmin=-maximum(abs.(plot_var1)), vmax = maximum(abs.(plot_var1)))
    
    pc1 = ax1.pcolormesh(plot_var1', cmap=PyPlot.cm.bwr, norm=norm3)

    colorbar(pc1)

    ax1.set_xlim(xl[1],xl[2])
    ax1.set_ylim(yl[1],yl[2])
    
    # plt.figure()
    plot_var2 = T_topo
    # plot_var2 = BCtrans
    # plot_var2 = BCprod
    # plot_var2 = u2

    norm4 = matplotlib.colors.TwoSlopeNorm(vcenter=0.,vmin=-maximum(abs.(plot_var2)), vmax = maximum(abs.(plot_var2)))
    
    # plt.contour(DEAPE', levels=[0.15e-9,1e-9], colors="black")   # for mu = 145^-1
    if cont_bool==true
        ax2.contour(plot_var1', levels=levs, colors="black")
    end

    
    pc2 = ax2.pcolormesh(plot_var2', cmap=PyPlot.cm.bwr, norm=norm4)   # PuOr
    
    colorbar(pc2)
    
    ax2.set_xlim(xl[1],xl[2])
    ax2.set_ylim(yl[1],yl[2])

    println(sum(plot_var1[xl[1]:xl[2],yl[1]:yl[2]]))

    println(sum(plot_var2[xl[1]:xl[2],yl[1]:yl[2]]))
    
end


levs = [-0.7e-10,0.7e-10]

mu = μ = (86400 * 45)^-1

# h_ind = 17

N_yr = 10



fig, ax = plt.subplots(1,2,figsize=(12,4))
ax1=ax[1]; ax2=ax[2];

plot_nrg_terms(mu, h_ind, N_yr, ax1, ax2, [1,512], [1,512], levs, false)
# plot_nrg_terms(mu, h_ind, N_yr, ax1, ax2, [1,200], [1,200], levs,false)

```

```julia
h_ind = 1


a = load(jld_name(data_dir,h0s[h_ind],Lx,rho,U,Nx,mu,H,N_yr))


ψtot = psi = a["jld_data"]["psi_ot"]

t_ind = 1

τ = 0.5 * (ψtot[:,:,1,t_ind] .- ψtot[:,:,2,t_ind])
ψ = 0.5 * (ψtot[:,:,1,t_ind] .+ ψtot[:,:,2,t_ind])

τh = rfft(τ)
ψh = rfft(ψ)

∂xτh = im * grid_jl.kr .* τh
∂xψh = im * grid_jl.kr .* ψh

∂yτh = im * grid_jl.l .* τh
∂yψh = im * grid_jl.l .* ψh

∂xτ = irfft(∂xτh, grid_jl.ny)
∂xψ = irfft(∂xψh, grid_jl.ny)

∂yτ = irfft(∂yτh, grid_jl.ny)
∂yψ = irfft(∂yψh, grid_jl.ny)

∂xψ2 = ∂xψ .- ∂xτ


BCprod = 0.01 * (28325)^-2 * τ .* ∂xτ

BCtrans = (f0/H[1]) * S32 * τ .* ∂xψ

T_topo = (f0/ 2 / H[2]) * h0s[h_ind] * τ .* ∂xψ2

T_topo_domain_avg = (f0/ 2 / H[2]) * h0s[h_ind] * τ .* ∂xψ

# ∇ψdot∇τ = ∂xτ .* ∂xψ .+ ∂yτ .* ∂yψ

# ∇ψ2 = ∂xψ.^2 .+ ∂yψ.^2

∇τ2 = ∂xτ.^2 .+ ∂yτ.^2

∇τdot∇ψ = ∂xτ .* ∂xψ .+ ∂yτ .* ∂yψ

DEAPE = -(μ/2) * (∇τ2 - ∇τdot∇ψ)

tot_trans = (f0/H[1] * S32 + f0 / 2 / H[2] * h0s[h_ind]) * τ .* ∂xψ2
```
